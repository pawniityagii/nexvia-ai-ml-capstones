## 1. Sample scans: normal vs abnormal


```python
fig,ax=plt.subplots(1,2,figsize=(8,4))
n,_=synthetic.generate_image(False,512,seed=1); a,bb=synthetic.generate_image(True,512,seed=2)
ax[0].imshow(n,cmap='gray'); ax[0].set_title('normal'); ax[0].axis('off')
ax[1].imshow(a,cmap='gray'); ax[1].set_title('abnormal (opacity)'); ax[1].axis('off'); plt.show()
```

## 2. Pixel-intensity distribution


```python
plt.hist(n.ravel(),bins=50,alpha=0.5,label='normal',density=True)
plt.hist(a.ravel(),bins=50,alpha=0.5,label='abnormal',density=True)
plt.legend(); plt.title('pixel intensity'); plt.show()
```

## 3. Grad-CAM-style heatmap localises the finding
The fallback saliency should light up the opacity. For the trained model, the
real Grad-CAM does the same (train_classifier_colab.ipynb).


```python
a,bb=synthetic.generate_image(True,512,seed=7)
r=fallback.predict(a/255.0)
fig,ax=plt.subplots(1,2,figsize=(8,4))
ax[0].imshow(a,cmap='gray'); ax[0].set_title('scan'); ax[0].axis('off')
ax[1].imshow(r['heatmap'],cmap='jet'); ax[1].set_title(f"saliency (p={r['probability']:.2f})"); ax[1].axis('off'); plt.show()
```

## Takeaways
- Opacity raises mean lung-field intensity; the model/heatmap should localise it.
- Use ROC-AUC + sensitivity, not accuracy, for screening.
